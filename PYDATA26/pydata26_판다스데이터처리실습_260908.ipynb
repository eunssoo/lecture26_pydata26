## 1. 파이썬 파일 처리

```python
sample_file = 'sample.txt'
```

```python
# 파일을 open하여 내용 출력
with open(sample_file, mode='r', encoding='utf-8') as f:
    content = f.read()
    print(content)
```

```python
# 파일을 open하고 한 줄씩 출력하기
with open(sample_file, mode='r', encoding='utf-8') as f:
    line = f.readline()

    while line:
        print(line, end='')
        line = f.readline()
```

```python
# 파일객체에서 한 줄씩 읽어 오기
# 파일 객체 f를 반복문에서 사용하면 파일을 한 줄씩 읽으며, 각 줄 끝의 \n도 포함해서 가져옵
with open(sample_file, mode='r', encoding='utf-8') as f:
    for line in f:
        print(line, end='')
```

```python
# 여러 개의 파일 읽기
import glob

files = glob.glob('data/*.txt')

for file_name in files:
    print(f'[{file_name}]')

    with open(file_name, mode='r', encoding='utf-8') as f:
        print(f.read())
```

## 2. 파이썬 문자열 다루기


```python
### 양쪽 공백 제거
b1=" 작은 일에 성실한 사람은      큰일에도 성실하고 "

print(b1)
print(b1.strip())
```

```python
# 왼쪽 문자, 오른쪽 문자, 양쪽 문자 제거
text="....작은 일에 성실한 사람은, 큰일에도 성실하고...."

print(text.lstrip('.'))
print(text.rstrip('.'))
print(text.strip('.'))
```

```python
# ,로 문자열 분리
text="....작은 일에 성실한 사람은, 큰일에도 성실하고...."

result = text.split(',')
print(result)

# 분리 후 각 문자열의 양쪽 '.'과 공백 제거
result = [item.strip().strip('.') for item in result]
print(result)
```

```python
# .으로 문자열 분리
numbers="1. 2. 3. 4. 5. 6"

result = numbers.split('.')
print(result)

# 공백 제거 후 빈 문자열 제외
result = [item.strip() for item in result if item.strip()]
print(result)
```
