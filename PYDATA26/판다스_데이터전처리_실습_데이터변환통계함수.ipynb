## 결측치

```python
import pandas as pd

# 타이타닉 데이터가 저장된 주소
url = 'https://raw.githubusercontent.com/sehakflower/data/main/titanic.csv'

# 데이터를 DataFrame으로 불러오기
titanic_df = pd.read_csv(url, sep='\t')

# 데이터 구조, 자료형, 결측치 개수 확인
titanic_df.info()
```

```python
# 사용할 컬럼 이름 지정
new_columns = ['passengerId', 'survived', 'pclass', 'name','sex', 'age', 'sibsp', 'parch','ticket', 'fare', 'cabin', 'embarked']
# 기존 컬럼 이름을 새 이름으로 변경
titanic_df.columns = new_columns
titanic_df.columns
```

```python
# age가 결측치(NaN)인 데이터만 추출
titanic_df['age'][titanic_df['age'].isna()]
```

```python
# age 열의 각 값이 결측치인지 확인
titanic_df['age'].isnull()
```

```python
# age가 결측치인 값만 출력
titanic_df['age'][titanic_df['age'].isnull()]
```

```python
# age 열의 결측치 개수 확인
titanic_df['age'].isnull().sum()
```

```python
# age 값이 존재하는지 확인
titanic_df['Age'].notnull()
```

```python
# age가 정상적으로 존재하는 데이터만 추출
titanic_df['age'][titanic_df['age'].notnull()]
```

```python
# age가 결측치가 아닌 행에서 age, name 열만 추출
titanic_df.loc[titanic_df['age'].notnull(),['age', 'name']] 
```

```python
# age가 결측치인 사람의 age와 name 출력
titanic_df.loc[titanic_df['Age'].isnull(),['Age', 'Name']]
```

```python
# 데이터 앞 5개 확인
titanic_df.head()
```

```python
# cabin의 NaN 값을 'N'으로 채우기
titanic_df['cabin'] = titanic_df['cabin'].fillna('N')

# 변경 결과 확인
titanic_df['cabin']     
```

```python
# cabin에 각 값이 몇 개 있는지 확인
titanic_df['cabin'].value_counts()
```

```python
# 0번 행의 cabin 값 확인
titanic_df['cabin'][0]
```

```python
# cabin 값에서 첫 번째 문자만 추출
titanic_df['cabin'] = titanic_df['cabin'].apply(lambda x: x[0])

# 결과 확인
titanic_df['cabin']
```

```python
# 전체 데이터 구조와 결측치 상태 확인
titanic_df.info()
```

```python
# 원본 DataFrame을 별도의 DataFrame으로 깊은 복사
copy_titanic_df = titanic_df.copy()

# 복사 결과 확인
copy_titanic_df
```

```python
# 원본 DataFrame의 메모리 주소
hex(id(titanic_df))
```

```python
# 복사본 DataFrame의 메모리 주소
hex(id(copy_titanic_df))
```

```python
# 원본 확인
titanic_df.head()
```

```python
# 복사본 확인
copy_titanic_df.head()
```

```python
# 복사본의 age 값을 모두 0으로 변경
copy_titanic_df['age'] = 0

# 복사본 확인
copy_titanic_df.head()
```

## 통계 함수

```python
# 숫자형 열의 주요 통계값을 한 번에 확인
titanic_df.describe()
```

```python
# 각 열의 값들을 모두 더함
titanic_df.sum()
```

```python
# 각 열에서 NaN이 아닌 데이터 개수 확인
titanic_df.count()
```

```python
# 각 열의 최댓값 확인
titanic_df.max()
```

```python
# 각 열의 최솟값 확인
titanic_df.min()
```

```python
# 숫자형 열의 평균값 구하기
# age 열은 숫자형이므로 정상 실행
titanic_df['age'].mean()
```

```python
# 숫자형 열의 중앙값 구하기
titanic_df.median(numeric_only=True)
```

```python
# 각 열에서 가장 많이 등장한 값 구하기
titanic_df.mode()
```

```python
# 평균 절대 편차 구하기
# age 열의 평균 절대 편차
(titanic_df['age'] - titanic_df['age'].mean()).abs().mean()
```

```python
# 숫자형 열의 분산 구하기
titanic_df.var(numeric_only=True)
```

```python
# 숫자형 열의 표준편차 구하기
titanic_df.std(numeric_only=True)
```

```python
# 숫자형 열의 왜도 구하기
titanic_df.skew(numeric_only=True)
```

```python
# 숫자형 열의 첨도 구하기
titanic_df.kurt(numeric_only=True)
```
